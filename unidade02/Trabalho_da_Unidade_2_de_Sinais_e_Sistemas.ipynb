## Parte 1
Plotar a resposta ao impulso e a resposta ao sinal $u(k)=1$ (com condição inicial nula) para o sistema LIT com o seguinte modelo de ED1:

$$
y(k+1)+0.9y(k)=2u(k)
$$

### Téoria:
#### **Modelo Padrão (MP):**
$$
x(k+1) + 0.9x(k) = u(k)
$$

Portanto, teremos que:
$$
y(k) = 2 \cdot x(k)
$$

#### **Resposta ao impulso - $h(k)$:**
$$
h(k) =
\begin{cases}
0.9^{k-1}, & x \geq 1, \\
0,  & x \geq 0.
\end{cases}
$$

#### **Resposta ao $y(k)$ com $u(k) = 1$:**
Para achar a resposta $y(k)$ se faz necessário achar primeiramente a respoda do modelo padrão. Desse modo:
$$
x(k) = x_o(k) + x_p(k) =
\begin{cases}
x_o(k) = 0.9^k \cdot x(0) \rightarrow \text{ Resposta Homogenea } \\
x_p(k) = (u * h)(k) \rightarrow \text{ Resposta Particular }
\end{cases}
$$

##### **Resposta Homogênea do MP - $x_o(k)$:**
Para calcular a resposta homogênea é necessário saber quem é $x(0)$, então:
$$
y(k) = 2\cdot x(k) \Rightarrow x(k) = \frac{y(k)}{2} \\
k = 0  \Rightarrow x(0) = \frac{y(0)}{2} = \frac{0}{2} = 0
$$

Portanto, teremos então que:
$$
x_o(k) = 0.9^k \cdot x(0) = 0.9^k \cdot 0 = 0
$$

##### **Resposta Particular - $x_p(k)$:**
$$x_p(k) = (u * h)(k) \longrightarrow^{z} X_P(z) = U(z) \cdot H(z)
\Rightarrow
\begin{cases}
U(z) = \frac{z}{z - 1} \\
H(z) = \frac{1}{z - 0.9}
\end{cases}
$$

Portanto, fazendo a convolução por Tranformada Z teremos:
$$
X_P(z) = 10 \cdot \frac{z}{z - 1} - 10 \cdot \frac{z}{z - 0.9}
$$

Desse modo, tirando do dominio de Z, teremos que:
$$
x_p(k) = 10 - 10 \cdot 0.9^k
$$

##### **Resposta completa do MP - $x(k)$:**
$$
x(k) = x_o(k) + x_p(k) = 0 + 10 - 10 \cdot 0.9^k = 10 - 10 \cdot 0.9^k
$$

Portanto $x(k) = 10 - 10 \cdot 0.9^k$.

##### **Resposta completa do modelo do sistema real - $y(k)$:**
$$
y(k) = 2 \cdot x(k) \Rightarrow y(k) = 2 \cdot (10 - 10 \cdot 0.9^k)
$$

Portanto, teremos que:
$$
y(k) = 20 - 20 \cdot 0.9^k
$$


### Plotando a resposta ao impulso $h(k)$

```python
# Bibliotecas necessárias
import matplotlib.pyplot as plt
import numpy as np

# Variaveis globais
t = 1
T_array = np.arange(0, 50, t)

# Gerando o sinal do impulso
h = np.power(0.9, T_array)

# Plotando sinal h(k)
plt.plot(T_array, h, 'o')
plt.title("Resposta ao sinal $y(k)$")
plt.xlabel("k")
plt.ylabel("y")
plt.vlines(T_array, 0, h)
plt.axhline(0, color='black')
plt.show()
```

## Parte 2
Plotar a resposta ao impulso e a resposta ao sinal $u(k)=1$ (com condição inicial nula) para o sistema LIT com o seguinte modelo de ED2:

$$
y(k+2) + (1.2a)y(k+1) + 0.36y(k)=2u(k)
$$

Considere os seguintes valores para o parâmetro "a": 0, 0.7, 1 e 1.5

### Teória:
#### **Modelo Padrão (MP):**
$$
x(k+2) + (1.2a)x(k+1) + 0.36x(k) = u(k)
$$

Portanto, teremos que:
$$
y(k) = 2 \cdot x(k) \\
h(k) = 2 \cdot h_x(k)
$$

Onde, teremos que:
$$
x(k) = x_o(k) + x_p(k) =
\begin{cases}
x_o(k) = h_x(k+1) * x(1) - a_0 \cdot h_x(k) \cdot x(0) \rightarrow \text{ Resposta Homogenea } \\
x_p(k) = (u * h_x)(k) \rightarrow \text{ Resposta Particular }
\end{cases}
$$

#### **Resposta ao impulso do MP:**
##### **Equação Característica:**
$$
x^2 + 1.2a \cdot x + 0.36 = 0 \Rightarrow
\begin{cases}
r_1 = -0.6a + 0.6\sqrt {a^2 - 1}\\
r_2 = -0.6a - 0.6\sqrt{a^2 - 1}
\end{cases}
$$

- Fazendo $a = 0$:
$$
\begin{cases}
r_1 = +0.6j\\
r_2 = -0.6j
\end{cases}
\Rightarrow
\begin{cases}
\alpha = 0\\
\omega = 0.6
\end{cases}
$$

- Fazendo $a = 0.7$:
$$
\begin{cases}
r_1 = -0.42 + 0.4285j\\
r_2 = -0.42 - 0.4285j
\end{cases}
\Rightarrow
\begin{cases}
\alpha = 0.42\\
\omega = 0.4285
\end{cases}
$$

- Fazendo $a = 1$:
$$
\begin{cases}
r_1 = -0.6\\
r_2 = -0.6
\end{cases}
\Rightarrow
r_1 = r_1 = r \in \mathbb{R}
$$

- Fazendo $a = 1.5$:
$$
\begin{cases}
r_1 = -0.3\\
r_2 = -1.5
\end{cases}
\Rightarrow
r_1 \neq r_2; ~ r_1, r_2 \in \mathbb{R}
$$

##### **Equação da resposta ao impulso:**
- Para $a = 0$:
$$
h_x(k) = \frac{R^{k-1}}{\omega} \cdot sen[(k-1) \cdot \Phi] =
\begin{cases}
R = \sqrt{\alpha^2 + \omega^2} = 0.6\\
\Phi = tg^{-1}(\frac{\omega}{\alpha}) = \frac{\pi}{2} \approx 1.571
\end{cases}
$$

Então:
$$
h_x(k) = \frac{0.6^{k-1}}{0.6} \cdot sen[(k-1) \cdot \frac{\pi}{2}] \longrightarrow^z H_x(z) = \frac{1}{z^2 + (0.6)^2}
$$

- Para $a = 0.7$:
$$
h_x(k) = \frac{R^{k-1}}{\omega} \cdot sen[(k-1) \cdot \Phi] =
\begin{cases}
R = \sqrt{\alpha^2 + \omega^2} = 0.6\\
\Phi = tg^{-1}(\frac{\omega}{\alpha}) = \approx 0.795
\end{cases}
$$

Então:
$$
h_x(k) = \frac{0.6^{k-1}}{0.4285} \cdot sen[(k-1) \cdot 0.795]
\longrightarrow^z H_x(z) = \frac{1}{(z - 0.42)^2 + (0.4285)^2}
$$

- Para $a = 1$:
$$
h_x(k) = (k-1)\cdot r^{k-2} = (k-1)\cdot (-0.6)^{k-2}
\longrightarrow^z H_x(z) = \frac{1}{(z + 0.6)^2}
$$

- Para $a = 1.5$:
$$
h_x(k) = \frac{r_1^{k-1} - r_2^{k-1}}{r_1 - r_2} = \frac{(-0.3)^{k-1} - (-1.5)^{k-1}}{-0.3 + 1.5} = \frac{(-0.3)^{k-1} - (-1.5)^{k-1}}{1.2}
$$

Logo:
$$
H_x(z) = \frac{1}{(z + 0.3)(z + 1.5)}
$$

#### **Resposta ao Impulso do modelo do sistema real:**
- Para $a = 0$:
$$
h(k) = 2 ⋅ h_x(k) = 2 \cdot \frac{0.6^{k-1}}{0.6} \cdot sen[(k-1) \cdot \frac{\pi}{2}]
$$

Portanto:
$$
h(k) = 3.333 \cdot 0.6^{k-1}\cdot sen[(k-1) \cdot \frac{\pi}{2}]
$$
- Para $a = 0.7$:
$$
h(k) = 2 \cdot h_x(k)
= 2 \cdot \frac{0.6^{k-1}}{0.4285} \cdot sen[(k-1)\cdot 0.795]
$$

Portanto:
$$
h(k) = 4.667 \cdot 0.6^{k-1} \cdot sen[(k-1)\cdot 0.795]
$$

- Para $a = 1$:
$$
h(k) = 2 \cdot h_x(k)
= 2\cdot (k-1)\cdot (-0.6)^{k-2}
$$

Portanto:
$$
h(k) = 2(k-1)(-0.6)^{k-2}
$$

- Para $a = 1.5$:
$$
h(k) = 2 \cdot h_x(k)
=
2\cdot
\frac{(-0.3)^{k-1}-(-1.5)^{k-1}}{1.2}
$$

Portanto:
$$
h(k) = \frac{5}{3} \left[(-0.3)^{k-1}-(-1.5)^{k-1}\right]
$$


### Plotando as Respostas ao Impulso

#### Para a = 0:
$$
h(k) = 3.333 \cdot 0.6^{k-1}\cdot sen[(k-1) \cdot \frac{\pi}{2}]
$$


```python
# Resposta ao impulso
h_a0 = 3.333 * np.power(0.6, T_array - 1) * np.sin((T_array - 1) * np.pi / 2)

# Plotando resposta ao impulso
plt.plot(T_array, h_a0, 'o')
plt.title("Resposta ao impulso $h(k)$")
plt.xlabel("k")
plt.ylabel("h(k)")
plt.vlines(T_array, 0, h_a0)
plt.axhline(0, color='black')
plt.show()
```

#### Para a = 0.7:
$$
h(k) = 4.667 \cdot 0.6^{k-1} \cdot sen[(k-1)\cdot 0.795]
$$

```python
# Resposta ao impulso
h_a07 = 4.667 * np.power(0.6, T_array - 1) * np.sin((T_array - 1) * 0.795)

# Plotando
plt.plot(T_array, h_a07, 'o')
plt.title("Resposta ao impulso $h(k)$")
plt.xlabel("k")
plt.ylabel("h(k)")
plt.vlines(T_array, 0, h_a07)
plt.axhline(0, color='black')
plt.show()
```

#### Para a = 1:
$$
h(k) = 2(k-1)(-0.6)^{k-2}
$$

```python
# Resposta ao impulso
h_a1 = 2 * (T_array - 1) * np.power(-0.6, T_array - 2)

# Plotando
plt.plot(T_array, h_a1, 'o')
plt.title("Resposta ao impulso $h(k)$")
plt.xlabel("k")
plt.ylabel("h(k)")
plt.vlines(T_array, 0, h_a1)
plt.axhline(0, color='black')
plt.show()
```

#### Para a = 1.5:
$$
h(k) = \frac{5}{3} \left[(-0.3)^{k-1}-(-1.5)^{k-1}\right]
$$

```python
# Resposta ao impulso
h_a15 = (5 / 3) * (np.power(-0.3, T_array - 1) - np.power(-1.5, T_array - 1))

# Plotando
plt.plot(T_array, h_a15, 'o')
plt.title("Resposta ao impulso $h(k)$")
plt.xlabel("k")
plt.ylabel("h(k)")
plt.vlines(T_array, 0, h_a15)
plt.axhline(0, color='black')
plt.show()
```
